```python
import os

from ultralytics import YOLO

model = YOLO("yolov8n.yaml")

results = model.train(data=os.path.join(ROOT_DIR, "data.yaml"), epochs=80, device = "cuda")
```

```python
best_model_path = os.path.join(results.save_dir, 'weights', 'best.pt')
best_model = YOLO(best_model_path)
```

```python
val_metrics = best_model.val(data=os.path.join(ROOT_DIR, 'data.yaml'), split='val')

print('Validation metrics:')
print(val_metrics)
```

```python
test_metrics = best_model.val(data=os.path.join(ROOT_DIR, 'data.yaml'), split='test')
print('\nTest metrics:')
print(test_metrics)
```
