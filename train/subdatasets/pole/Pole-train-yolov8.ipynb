```python
import os
from ultralytics import YOLO

# Load a pretrained YOLOv8n model
model = YOLO("yolov8n.yaml")

# Train the model
results = model.train(data=os.path.join(ROOT_DIR, "data.yaml"), epochs=80, device="cuda")
```

### Validation 1

```python
best_model_path = os.path.join(results.save_dir, 'weights', 'best.pt')
best_model = YOLO(best_model_path)

val_metrics = best_model.val(data=os.path.join(ROOT_DIR, 'data.yaml'), split='val', name='val')

print('Validation metrics:')
print(val_metrics)
```

### Validation 2 (Optional)

```python
# Run this again if you want a second validation folder (val2)
val_metrics2 = best_model.val(data=os.path.join(ROOT_DIR, 'data.yaml'), split='val', name='val2')
print(val_metrics2)
```
